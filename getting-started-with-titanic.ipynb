```python
training_data = pd.read_csv("/kaggle/input/titanic/train.csv")
training_data.head()
```

```python
test_data = pd.read_csv("/kaggle/input/titanic/test.csv")
test_data.head()
```

```python
women = training_data.loc[training_data.Sex == 'female']["Survived"]
rate_women = sum(women)/len(women)

print("% of women who survived:", rate_women)
```

```python
men = training_data.loc[training_data.Sex=="male"]["Survived"]
rate_men = sum(men)/len(men)
print("% of men who survived:" , rate_men)
```

```python
from sklearn.ensemble import RandomForestClassifier
features = ["Pclass", "Sex", "SibSp", "Parch", "Fare", "Embarked"]
combined = pd.concat([training_data[features], test_data[features]])
combined["Fare"] = combined["Fare"].fillna(combined["Fare"].median())
combined["Embarked"] = combined["Embarked"].fillna(combined["Embarked"].mode()[0])

combined = pd.get_dummies(combined)
train_x = combined[:len(training_data)]
train_y = training_data["Survived"]
test_x  = combined[len(training_data):]

model = RandomForestClassifier(n_estimators = 100, random_state =1, max_depth = 5 )
model.fit(train_x, train_y)

predictions = model.predict(test_x)
output = pd.DataFrame({"PassengerId": test_data["PassengerId"],
    "Survived": predictions})

output.to_csv('submission.csv', index=False)
print("Your submission was successfully saved!")

```
